#### **Handling Duplicate Row**

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

#### **Handling Outlier**

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

outlier_mask = (df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))
outlier_mask.sum().sort_values(ascending=False).head(10)
```

```python
df_clean = df.copy()

# hitung IQR hanya untuk fitur mean
Q1 = df_clean[features].quantile(0.25)
Q3 = df_clean[features].quantile(0.75)
IQR = Q3 - Q1

# mask data TANPA outlier
mask = ~(
    (df_clean[features] < (Q1 - 1.5 * IQR)) |
    (df_clean[features] > (Q3 + 1.5 * IQR))
).any(axis=1)

# drop outlier
df_clean = df_clean[mask]

df_clean.shape
```

#### **Scaling**

```python
numerical_cols = df.select_dtypes(include=['number']).columns
target_col = 'Class'

numerical_cols = numerical_cols.drop(target_col)

scaler = StandardScaler()
df[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```
